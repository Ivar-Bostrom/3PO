```python
def empty_schema(spark, catalog: str, schema: str):
    full_schema = f"`{catalog}`.`{schema}`"
    tables = spark.sql(f"SHOW TABLES IN {full_schema}").collect()
    volumes = spark.sql(f"SHOW VOLUMES IN {full_schema}").collect()
    for row in tables:
        table_name = row["tableName"]
        stmt = f"DROP TABLE IF EXISTS `{catalog}`.`{schema}`.`{table_name}`"
        spark.sql(stmt)
    for row in volumes:
        volume_name = row["volume_name"]
        stmt = f"DROP VOLUME IF EXISTS `{catalog}`.`{schema}`.`{volume_name}`"
        spark.sql(stmt)

def empty_table(spark, catalog: str, schema: str, table: str):
    full_name = f"{catalog}.{schema}.{table}"
    spark.sql(f"DELETE FROM {full_name}")

```

```python
empty_schema(spark, "3po","main")
empty_schema(spark, "3po","tm")
empty_schema(spark, "3po","tb")
empty_schema(spark, "3po","cm")
empty_schema(spark, "3po","cb")
empty_schema(spark, "3po","hm")
empty_schema(spark, "3po","hb")
```
