```python
%sql
CREATE TABLE IF NOT EXISTS 3po.main.timeline (
time INTEGER
)
USING DELTA
TBLPROPERTIES (
  'delta.appendOnly' = 'true',
  'delta.columnMapping.mode' = 'name',
  'delta.minReaderVersion' = '2',
  'delta.minWriterVersion' = '5'
);

CREATE TABLE IF NOT EXISTS 3po.main.ingests (
time INTEGER,
instances INTEGER,
posts INTEGER,
files INTEGER
)
USING DELTA
TBLPROPERTIES ('delta.appendOnly' = 'true');


CREATE TABLE IF NOT EXISTS 3po.main.bluesky_positivity (
  ID        STRING,
  positivity FLOAT      
)
USING DELTA
--TBLPROPERTIES ('delta.appendOnly' = 'true')
;

CREATE TABLE IF NOT EXISTS 3po.main.mastodon_positivity (
  ID        STRING,
  positivity FLOAT      
)
USING DELTA
--TBLPROPERTIES ('delta.appendOnly' = 'true')
;
```
