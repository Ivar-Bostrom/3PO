```python
def find_table_with_multiple_rows() -> str | None:
    tables = spark.sql(f"SHOW TABLES IN 3po.hbluesky").collect()

    for row in tables:
        table_name = row["tableName"]
        full_name = f"3po.hbluesky.`{table_name}`"

        try:
            # Use LIMIT 2 so Spark stops scanning as early as possible
            count = spark.sql(f"SELECT COUNT(*) FROM (SELECT 1 FROM {full_name} LIMIT 2)").collect()[0][0]

            if count > 1:
                print(f"Found: {full_name}")
            else:
                print(f"Skipping {full_name} ({count} row(s))")

        except Exception as e:
            print(f"Could not read {full_name}: {e}")
            continue

    print("No table with multiple rows found.")
    return None
find_table_with_multiple_rows()
```
